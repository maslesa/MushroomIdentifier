# Dataset split check

### 01. Import libraries

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from tqdm import tqdm
```

### 02. Load datasets

```python
DATASET_DIR = Path('../dataset')
TRAIN_DIR = DATASET_DIR / 'train'
VAL_DIR = DATASET_DIR / 'validation'
TEST_DIR = DATASET_DIR / 'test'
```

#### 02.01. Function for loading dataset

```python
def load_dataset(split_dir: Path) -> pd.DataFrame:
    data = []

    for species_dir in sorted(split_dir.iterdir()):
        if not species_dir.is_dir():
            continue

        for image_path in sorted(species_dir.iterdir()):
            if image_path.suffix.lower() not in {'.jpg', '.png'}:
                continue

            data.append({
                'species': species_dir.name,
                'filename': image_path.name,
                'path': image_path
            })

    return pd.DataFrame(data)
```

```python
train_df = load_dataset(TRAIN_DIR)
val_df = load_dataset(VAL_DIR)
test_df = load_dataset(TEST_DIR)
```

#### 02.02. Datasets summary

```python
summary = pd.DataFrame({
    'Split': ['Train', 'Validation', 'Test'],
    'Number of species': [
        train_df['species'].nunique(),
        val_df['species'].nunique(),
        test_df['species'].nunique()
    ],
    'Number of images': [
        len(train_df),
        len(val_df),
        len(test_df)
    ]
})

summary
```

### 03. Class distribution

```python
distribution = pd.DataFrame({
    'Train': train_df['species'].value_counts(),
    'Validation': val_df['species'].value_counts(),
    'Test': test_df['species'].value_counts()
}).fillna(0).astype(int)

distribution
```

### 04. Verify split ratio

```python
ratios = distribution.copy()

totals = ratios['Train'] + ratios['Validation'] + ratios['Test']

ratios['Train %'] = round(ratios['Train'] / totals * 100, 2)
ratios['Validation %'] = round(ratios['Validation'] / totals * 100, 2)
ratios['Test %'] = round(ratios['Test'] / totals * 100, 2)

ratios
```
